```python
import os
import sys
os.chdir(r"D:\CO2_Masterarbeit\CO2_Masterarbeit")
sys.path.insert(0, r"D:\CO2_Masterarbeit\CO2_Masterarbeit\FIG")

import geopandas as gpd
import matplotlib.pyplot as plt
import pandas as pd
from plotting.apa7 import set_apa7_style, new_figure, save_apa7, BAR_KWARGS, HIST_KWARGS, LINE_KWARGS

set_apa7_style()
```

```python
schools = gpd.read_file("OGD/Bildungsstandorte.zip")
print(f"Sites: {len(schools):,}")
schools[["TYP_LANG", "BEZNAM", "PLZ"]].describe(include="all")
```

## Institution types

Number of recorded sites by institutional type.

```python
counts = schools["TYP_LANG"].value_counts().sort_values()
fig, ax = new_figure(7, 4.5)
ax.barh(counts.index, counts.values, **BAR_KWARGS)
ax.set(xlabel="Anzahl der Standorte", ylabel="")
save_apa7(fig, "FIG/Bildungsstandorte/figure_1_institution_types.png")
save_apa7(fig, "FIG/Bildungsstandorte/figure_1_institution_types.svg")
plt.show()
```

## Sites by district

```python
counts = schools["BEZNAM"].value_counts().sort_values()
fig, ax = new_figure(7, 6)
ax.barh(counts.index, counts.values, **BAR_KWARGS)
ax.set(xlabel="Anzahl der Standorte", ylabel="")
save_apa7(fig, "FIG/Bildungsstandorte/figure_2_sites_by_district.png")
save_apa7(fig, "FIG/Bildungsstandorte/figure_2_sites_by_district.svg")
plt.show()
```
