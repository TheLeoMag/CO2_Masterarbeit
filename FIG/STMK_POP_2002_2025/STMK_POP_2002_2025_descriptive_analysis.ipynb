```python
population = pd.read_csv("OGD/STMK_POP_2002_2025.csv", sep=";", encoding="cp1252")
population["POP_2025"].describe()
```

## Statewide population trend

Annual population is summed across municipalities.

```python
years = [f"POP_{year}" for year in range(2002, 2026)]
totals = population[years].sum()
fig, ax = new_figure()
ax.plot(range(2002, 2026), totals, **LINE_KWARGS)
ax.set(xlabel="Jahr", ylabel="Bevölkerung")
ax.ticklabel_format(axis="y", style="plain")
save_apa7(fig, "FIG/STMK_POP_2002_2025/figure_1_statewide_population_trend.png")
save_apa7(fig, "FIG/STMK_POP_2002_2025/figure_1_statewide_population_trend.svg")
plt.show()
```

## Municipal population in 2025

```python
fig, ax = new_figure()
ax.hist(population["POP_2025"], bins=35, **HIST_KWARGS)
ax.set_xscale("log")
ax.set(xlabel="Gemeindebevölkerung 2025 (logarithmische Skala)", ylabel="Häufigkeit")
save_apa7(fig, "FIG/STMK_POP_2002_2025/figure_2_municipal_population_2025.png")
save_apa7(fig, "FIG/STMK_POP_2002_2025/figure_2_municipal_population_2025.svg")
plt.show()
```
